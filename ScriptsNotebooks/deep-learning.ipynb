# Treinamento Deep Learning

# Variáveis X e y

```python
#variaveis independetes X e variavel dependente y
import pandas as pd
dataset = pd.read_csv('/kaggle/input/datasets/adessz/churn-modelling/Churn_Modelling.csv')
X = dataset.iloc[:, 3:13].values
y = dataset.iloc[:, 13].values
```

# Dados categóricos e númericos

```python
#transformar dados categoricos em dados numericos
from sklearn.preprocessing import LabelEncoder, OneHotEncoder
from sklearn.compose import make_column_transformer
labelencoder_X_1 = LabelEncoder()
X[:, 1] = labelencoder_X_1.fit_transform(X[:, 1])
labelencoder_X_2 = LabelEncoder()
X[:, 2] = labelencoder_X_2.fit_transform(X[:, 2])
```

# Variável dummy

```python
#variavel dummie para transformar os valores de país que tem mais de duas categorias em 3 colunas
onehotencoder= make_column_transformer((OneHotEncoder(categories='auto', sparse_output = False), [1]), remainder="passthrough")
X=onehotencoder.fit_transform(X)
X = X[:,1:]
```

# Treino e Teste

```python
#dividir o conjunto de dados em treino e teste
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 0)
```

# Padronização Feature Scalling

```python
#feature scalling para padronizar a escala dos valores númericos
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

# Rede Neural

```python
#criar a rede neural
from tensorflow import keras
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

classifier = Sequential() #inicia a rede neural
```

# Camadas Densas

```python
#criando camadas densas: units 6 numero de neuronios na camada oculta, input_dim numero de neuronios de entradas que sao os 11 atributos,
#o numero de neuronio units pode ser a media entre camada de entrada 11 e c saida 1 = 6
classifier.add(Dense(6, kernel_initializer = 'uniform', activation = 'relu', input_shape = (11,))) 
classifier.add(Dense(units = 6, kernel_initializer = 'uniform', activation = 'relu'))
classifier.add(Dense(units = 1, kernel_initializer = 'uniform', activation = 'sigmoid')) #camada de saída
```

# Otimizador

```python
#adam otimizador(ajustar os pesos) baseado em sthocastic gradient descent, loss_function = funçao de perda binaria
classifier.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics = ['accuracy'])
```

# Previsão

```python
#fazendo a previsao
classifier.fit(X_train, y_train, batch_size = 10, epochs = 10)
y_pred = classifier.predict(X_test)
y_pred = (y_pred > 0.5)
```

# Matriz de Confusão

```python
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test, y_pred)
cm
```
