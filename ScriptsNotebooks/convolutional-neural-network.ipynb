# Treinamento Convolutional Neural Network

# Bibliotecas

```python
#importando bibliotecas
from tensorflow import keras
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense
```

```python
#inicializando a rede
classifier = Sequential()
```

# Camada Convolucional

```python
#criando a camada convolucional com parametros de filtros 32 (3x3), formato da imagem(64,64,3),
#função de ativação 
classifier.add(Conv2D(32,(3,3),input_shape = (64, 64, 3), activation = 'relu'))
```

# MaxPooling

```python
#dividindo o mapa de características em janelas 2x2 selecionando apenas o maior valor.
classifier.add(MaxPooling2D(pool_size = (2, 2)))
```

```python
#criando mais uma camada convolucional com pooling para melhorar o resultado dos testes
classifier.add(Conv2D(32, (3, 3), activation = 'relu'))
classifier.add(MaxPooling2D(pool_size = (2, 2)))
```

# Flatten

```python
#matriz de características é convertida em um vetor de caracteristicas
classifier.add(Flatten())
```

# Camadas Densas

```python
#criando as camadas densas, a rede neural.
#camada de entrada com o número de neuronios
classifier.add(Dense(128, activation = 'relu'))
#camada de saída = 1 porque a saída é binária. Sigmóide pela resposta percentual.
classifier.add(Dense(1, activation = 'sigmoid'))
#compilar usando o otimizador adam baseado no gradient descent, funcao de perda e 
#ajuste dos pesos, métrica de avaliaçao
classifier.compile(optimizer = 'adam', loss = 'binary_crossentropy', metrics = ['accuracy']) 
```

# Pré processamento IMG

```python
from tensorflow.keras.preprocessing.image import ImageDataGenerator
train_datagen = ImageDataGenerator(rescale = 1./255,
                                   shear_range = 0.2,
                                   zoom_range = 0.2,
                                   horizontal_flip = True)
test_datagen = ImageDataGenerator(rescale = 1./255)
```

# Treino e Teste

```python
#carregando os dados de treino, local, tamanho da imagem, o tamanho dos batchs lotes,classe binária
training_set = train_datagen.flow_from_directory('dataset/training_set',
                                                 target_size = (64, 64),
                                                 batch_size = 32,
                                                 class_mode = 'binary')

#carregando os dados de teste, local, tamanho da imagem, o tamanho dos batchs lotes,classe binária
test_set = test_datagen.flow_from_directory('dataset/test_set',
                                            target_size = (64, 64),
                                            batch_size = 32,
                                            class_mode = 'binary')
```

# Validação

```python
#treinado a rede, steps_per_epoch= 8000 numero de passos por epoch igual ao conjunto de treino,
#a ideia é igual para validation_steps = 2000
classifier.fit_generator(training_set,steps_per_epoch = 8000,
epochs = 25,validation_data = test_set,validation_steps = 2000)
```
