# Machine Learning

```python
#Gerando Dados para Machine Learning Clássico (Scikit-Learn)
from sklearn.datasets import make_classification
import pandas as pd
```

```python
# Cria 100 linhas, 4 colunas de características (features) e 1 coluna de resposta (0 ou 1)
X, y = make_classification(n_samples=100, n_features=4, random_state=42)
```

```python
# Transforma em DataFrame
df = pd.DataFrame(X, columns=['feat_1', 'feat_2', 'feat_3', 'feat_4'])
df['alvo'] = y
```

```python
print(df.head())
```

# Imagens Nativas para Visão Computacional (OpenCV)

```python
#Carregando Imagens Nativas para Visão Computacional (OpenCV)
import cv2
import numpy as np
```

```python
# Cria uma imagem preta de 400x400 pixels direto na memória (3 canais de cor: BGR)
imagem_teste = np.zeros((400, 400, 3), dtype=np.uint8)
```

```python
# Desenha um círculo branco no meio dessa imagem para testar um filtro
cv2.circle(imagem_teste, (200, 200), 50, (255, 255, 255), -1)
```

```python
# Aplica um filtro de desfoque (Blur) para provar que o OpenCV está funcionando
imagem_borrada = cv2.GaussianBlur(imagem_teste, (15, 15), 0)
```
